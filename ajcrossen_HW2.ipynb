### Question 1b

sure to not include incidents that occurred in Iowa City, CA. Save results to a new

### Question 1E
In 2016, which cities and states had the most fatalities? Display the maximum 
number killed in 2016. Show n_killed, city_or_county, and state columns for 
records that match the maximum number killed.

```python
# convert the date column to a date field
gun_violence["date"] = pd.to_datetime(gun_violence["date"])

# create a new DataFrame containing only incidents from 2016
incidents_2016 = gun_violence[
    gun_violence["date"].dt.year == 2016
]

# group by city_or_county and state
# then sum the groups by n_killed
aggregated_2016 = incidents_2016.groupby(
    ["city_or_county", "state"],
    as_index=False
)["n_killed"].sum()

# find the maximum number killed
max_killed = aggregated_2016["n_killed"].max()

# find all cities and states that match the maximum
most_fatalities_2016 = aggregated_2016[
    aggregated_2016["n_killed"] == max_killed
]

# display results with narrative context
print(f"Maximum number killed in 2016: {max_killed}")
display(
    most_fatalities_2016[
        ["n_killed", "city_or_county", "state"]
    ]
)
```

### Question 2a

Subset the six specified player IDs into a new data frame named `selected_players`.
Keep only the `sofifa_id` and `long_name` columns.

```python
#load the FIFA 21 dataset using UTF-8 encoding
fifa_21 = pd.read_csv("players_21.csv", encoding="utf-8")

#subset the six specified players
selected_players = fifa_21[
    fifa_21["sofifa_id"].isin([
        231747,
        153079,
        162835,
        200104,
        223848,
        208421
    ])
][["sofifa_id", "long_name"]]

# display the selected players
print("Selected FIFA 21 players:")
display(selected_players)
```

### Question 2b

Subset players with a height greater than or equal to 191 cm
and a weight greater than or equal to 90 kg into a new data frame named `big_players`.
Display the number of players in the new data frame.

```python
# subset players who meet both height and weight criteria
big_players = fifa_21[
    (fifa_21["height_cm"] >= 191) &
    (fifa_21["weight_kg"] >= 90)
]

# display the number of players who meet the criteria
print(f"There are {len(big_players)} players who meet the criteria.")
```

### Question 2c

Determine the oldest player age, youngest player age, median age,
and mean age of all players. Store each result in a variable and
display the results as whole numbers.

```python
#calculate age statistics
oldest_age = fifa_21["age"].max()
youngest_age = fifa_21["age"].min()
median_age = fifa_21["age"].median()
mean_age = fifa_21["age"].mean()

#display the results as whole numbers
print(f"Oldest player age: {round(oldest_age)}")
print(f"Youngest player age: {round(youngest_age)}")
print(f"Median player age: {round(median_age)}")
print(f"Mean player age: {round(mean_age)}")
```

### Question 2d

Show the long name, age, and height for the oldest player(s)
who prefer their left foot.z

```python
# find the oldest age among players who prefer their left foot
oldest_left_age = fifa_21[
    fifa_21["preferred_foot"] == "Left"
]["age"].max()

# filter for the oldest left-footed player(s)
oldest_left_players = fifa_21[
    (fifa_21["preferred_foot"] == "Left") &
    (fifa_21["age"] == oldest_left_age)
][["long_name", "age", "height_cm"]]

# display the result
print("Oldest left-footed player(s):")
display(oldest_left_players)
```
